Problem statement:- Python for Data Handling Select dataset from https://www.kaggle.com/datasets and do Data Visualization using Pandas. Plot Histogram, Boxplot, Scatterplot and Bar chart.

Dataset Selected: Iris Flower Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Iris.csv")
df = pd.DataFrame(data)
print(df.head())
print(df.info())
```

### 1. Histogram

```python
df['SepalLengthCm'].hist(bins=20)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

### 2. Boxplot

```python
df.boxplot(column='PetalLengthCm')
plt.title("Boxplot of Petal Length")
plt.ylabel("Petal Length (cm)")
plt.show()
```

### 3. Scatter Plot

```python
df.plot.scatter(x='SepalLengthCm', y='PetalLengthCm')
plt.title("Sepal Length vs Petal Length")
plt.show()
```

### 4. Bar Chart

```python
df['Species'].value_counts().plot(kind='bar')
plt.title("Count of Each Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```
